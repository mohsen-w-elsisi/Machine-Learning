### Defining Performance Metrics 

#### Mean Squared Error

- Sensitive to outliers
- Changes units

$$
MSE = \frac{1}{n} \sum^n_{i=1}{(y_i-\hat{y}_i)^2}
$$

```python
def MSE(y_true, y_pred):
    return ((y_true - y_pred) ** 2).mean()
```

#### Root Mean Squared Error

- sensetive to outliers (inheiriting from quadratic nature of MSE)
- Keeps same units as data 

$$
RMSE = \sqrt{MSE}
$$

```python
def RMSE(y_true, y_pred):
    return MSE(y_true, y_pred) ** 0.5
```

#### Mean Absolute Error

- cannot be used for optimisation due to absolute function
- keeps real world units
- resistant to outliers (linear function)

$$
\frac{1}{n} \sum^n_{i=1}{|y_i-\hat{y}_i|}
$$

```python
def MAE(y_true, y_pred):
    return (y_true - y_pred).abs().sum() / len(y_true)
```

### Testing the Models

```python
from pandas import DataFrame, Series


def model_test_results(predictions):
    return Series(
        {
            "MSE": MSE(testing_data_y, predictions),
            "RMSE": RMSE(testing_data_y, predictions),
            "MAE": MAE(testing_data_y, predictions),
        }
    )
```

```python
DataFrame(
    [
        model_test_results(gd_regresser.predict(testing_data_features)),
        model_test_results(dm_regresser.predict(testing_data_features)),
    ],
    index=["Gradient Descent", "Direct Method"],
)
```

The difference is between the two approaches of training is unarguably neglagable, however the direct minimisation approach took significantly less time. This can be attributed to the small number of features, meaning only a 3x3 matrix needed to be inverted.
